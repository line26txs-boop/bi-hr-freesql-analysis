```python
import pandas as pd

df_salarios = pd.read_csv('../data/query_01.csv')
df_regioes = pd.read_csv('../data/query_02.csv')

print("df_salarios:", df_salarios.shape)
print("df_regioes:", df_regioes.shape)
```

```python
print("=== SHAPE ===")
print("Salários:", df_salarios.shape)
print("Regiões:", df_regioes.shape)

print("\n=== INFO: df_salarios ===")
df_salarios.info()

print("\n=== INFO: df_regioes ===")
df_regioes.info()
```

```python
print("=== DESCRIBE: df_salarios ===")
display(df_salarios.describe())

print("\n=== DESCRIBE: df_regioes ===")
display(df_regioes.describe())
```

```python
print("=== NULOS: df_salarios ===")
print(df_salarios.isnull().sum())

print("\n=== NULOS: df_regioes ===")
print(df_regioes.isnull().sum())
```

```python
print("Duplicados em df_salarios:", df_salarios.duplicated().sum())
print("Duplicados em df_regioes:", df_regioes.duplicated().sum())
```

```python
print("=== Análise exploratória ===")
print("\nTop 5 maiores salários (df_salarios):")
display(df_salarios.nlargest(5, 'SALARY'))

print("\nMédia salarial por departamento:")
display(df_salarios.groupby('DEPARTMENT_NAME')['SALARY'].mean().sort_values(ascending=False))

print("\nQuantidade de funcionários por região:")
display(df_regioes['REGION_NAME'].value_counts())
```

```python
print("=== ESTATÍSTICAS DESCRITIVAS: SALARY (df_salarios) ===")
media = df_salarios['SALARY'].mean()
mediana = df_salarios['SALARY'].median()
minimo = df_salarios['SALARY'].min()
maximo = df_salarios['SALARY'].max()
desvio = df_salarios['SALARY'].std()

print(f"Média:         R$ {media:,.2f}")
print(f"Mediana:       R$ {mediana:,.2f}")
print(f"Mínimo:        R$ {minimo:,.2f}")
print(f"Máximo:        R$ {maximo:,.2f}")
print(f"Desvio padrão: R$ {desvio:,.2f}")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.hist(df_salarios['SALARY'], bins=10, color='#4C72B0', edgecolor='black')
plt.title('Distribuição de Salários')
plt.xlabel('Salário (R$)')
plt.ylabel('Quantidade de Funcionários')
plt.tight_layout()
plt.savefig('../images/histograma_salarios.png', dpi=150)
plt.show()
```

```python
plt.figure(figsize=(6, 5))
plt.boxplot(df_salarios['SALARY'], vert=True)
plt.title('Boxplot de Salários')
plt.ylabel('Salário (R$)')
plt.tight_layout()
plt.savefig('../images/boxplot_salarios.png', dpi=150)
plt.show()
```

```python
media_dept = df_salarios.groupby('DEPARTMENT_NAME')['SALARY'].mean().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
plt.bar(media_dept.index, media_dept.values, color='#55A868')
plt.title('Salário Médio por Departamento')
plt.xlabel('Departamento')
plt.ylabel('Salário Médio (R$)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('../images/barras_departamento.png', dpi=150)
plt.show()
```

```python
contagem_regiao = df_regioes['REGION_NAME'].value_counts()

plt.figure(figsize=(7, 5))
plt.bar(contagem_regiao.index, contagem_regiao.values, color='#C44E52')
plt.title('Quantidade de Funcionários por Região')
plt.xlabel('Região')
plt.ylabel('Quantidade de Funcionários')
plt.tight_layout()
plt.savefig('../images/barras_regiao.png', dpi=150)
plt.show()
```
